## 1.5 New columns and sorting  (14 pts)
Work on a copy: `df2 = df.copy()`. Add two columns to `df2`:
- `ec_flag`: `1` if `Extracurricular Activities` is `"Yes"`, otherwise `0` (`map`)
- `study_group`: `"low"` if `Hours Studied` is 3 or less, `"medium"` if 4 to 6, `"high"` if 7 or more (`pd.cut`, `np.select`, or `apply`)

Then store `top5`: the 5 rows with the highest `Performance Index` (`nlargest`).

```python
df2 = df.copy()

df2["ec_flag"] = df2["Extracurricular Activities"].map({"Yes": 1, "No": 0})

import numpy as np
hrs = df2["Hours Studied"]
df2["study_group"] = np.select(
    [hrs <= 3, (hrs >= 4) & (hrs <= 6), hrs >= 7], 
    ["low", "medium", "high"],
    default="medium"
)

top5 = df2.nlargest(5, "Performance Index")
```

## 2.2 Impossible values and outliers  (15 pts)
`Previous Scores` should be between 0 and 100, but some rows have values like 400.
1. `n_invalid`: how many values in `df_dirty["Previous Scores"]` are **greater than 100**
2. Use the **IQR method** on `df_dirty["Previous Scores"]`: `q1`, `q3`, `iqr` (`q3 - q1`), `upper` (`q3 + 1.5 * iqr`), and `n_iqr_outliers`, the number of values above `upper`
3. Create `df_work = df_dirty.copy()` and **replace the impossible values with `NaN`** in `df_work` only

```python
n_invalid = (df_dirty["Previous Scores"] > 100).sum()

q1 = df_dirty["Previous Scores"].quantile(0.25)
q3 = df_dirty["Previous Scores"].quantile(0.75)
iqr = q3 - q1
upper = q3 + 1.5 * iqr
n_iqr_outliers = (df_dirty["Previous Scores"] > upper).sum()

df_work = df_dirty.copy()
df_work.loc[df_work["Previous Scores"] > 100, "Previous Scores"] = np.nan

print(n_invalid,'\n',upper,'\n' ,n_iqr_outliers)
```

```python
def t():
    assert n_invalid == 5
    assert abs(iqr - (q3 - q1)) < 1e-9 and abs(upper - (q3 + 1.5 * iqr)) < 1e-9
    assert n_iqr_outliers == 5, f"expected 5 IQR outliers, got {n_iqr_outliers}"
    assert df_work["Previous Scores"].isnull().sum() == 55, "Previous Scores should now have 50 + 5 missing values"
    assert df_dirty["Previous Scores"].max() == 400, "df_dirty must stay unchanged"
check2("2.2 impossible values and IQR", 15, t)
```
